```python
import matplotlib.pyplot as plt
import numpy as np

data = {
    '32': {'120': [0.312154, 0.312062, 0.312436, 0.312070, 0.312997],
           '240': [2.579642, 2.589721, 2.603955, 2.590127, 2.593110]},
    '48': {'120': [0.314334, 0.315050, 0.315425, 0.317775, 0.317313],
           '240': [2.562062, 2.575325, 2.574730, 2.588488, 2.591811]},
    '64': {'120': [0.310719, 0.310269, 0.310548, 0.311327, 0.311673],
           '240': [2.558982, 2.556734, 2.569651, 2.568102, 2.562984]},
    '96': {'120': [0.324966, 0.324048, 0.321586, 0.325786, 0.323346],
           '240': [2.577298, 2.592596, 2.573855, 2.572369, 2.566895]},
}

processes = ['32', '48', '64', '96']
grid_120 = [data[p]['120'] for p in processes]
grid_240 = [data[p]['240'] for p in processes]


fig, axes = plt.subplots(1, 2, figsize=(12, 5))

positions = np.arange(len(processes)) + 1  # [1,2,3,4]

# Common styling
common_props = dict(
    patch_artist=True,
    medianprops=dict(color='black', linewidth=2),
    whiskerprops=dict(linewidth=1.5),
    capprops=dict(linewidth=1.5),
    flierprops=dict(marker='o', markersize=4, alpha=0.5)
)

axes[0].boxplot(
    grid_120,
    positions=positions,
    widths=0.5,
    boxprops=dict(facecolor='#4C72B0', alpha=0.7),
    **common_props
)

axes[0].set_title(r'$120^3$ Grid')
axes[0].set_xlabel('Processes (P)')
axes[0].set_ylabel('Execution Time (s)')
axes[0].set_xticks(positions)
axes[0].set_xticklabels(processes)
axes[0].grid(axis='y', linestyle='--', alpha=0.4)

axes[1].boxplot(
    grid_240,
    positions=positions,
    widths=0.5,
    boxprops=dict(facecolor='#DD8452', alpha=0.7),
    **common_props
)

axes[1].set_title(r'$240^3$ Grid')
axes[1].set_xlabel('Processes (P)')
axes[1].set_xticks(positions)
axes[1].set_xticklabels(processes)
axes[1].grid(axis='y', linestyle='--', alpha=0.4)

for ax in axes:
    ax.set_facecolor('white')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.suptitle('Parallel Execution Time vs Processes', fontsize=14)
plt.tight_layout()
plt.savefig('clean_subplot_boxplot.png', dpi=300)
plt.show()
```
